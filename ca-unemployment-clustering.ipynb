### Section 5.3 KMeans Clustering

First, I'll create a new dataframe called county_df that filters the column "Area Type" from the original df to only show counties.

```python
county_df = df[df["Area Type"] == "County"]
county_df
```

Second, I'll create a groupby using the county_df I just created to groupby "Area Name" and by means to show each of the counties' means in numeric columns over the years.

```python
county_means = county_df.groupby("Area Name").mean(numeric_only=True).reset_index()
county_means
```

Now, that I've done all the filtering of the dataframes to show each respective county and their mean unmeployment rate in the 2007 to 2024 timeframe, I can move onto the Kmeans clustering. First, I'll import Kmeans from sklearn.cluster. Next, I'll train my model with 2 clusters then fit the model with "Unemployment Rate", "Employment" and "Unemployment" since the more variables I have, the better my model will be. Lastly, I'll make a column called "Cluster" with model.predict to get the final result of predicting "Unemployment Rate"

```python
from sklearn.cluster import KMeans

model = KMeans(2)

model.fit(county_means[["Unemployment Rate"]])

county_means["Cluster"] = model.predict(county_means[["Unemployment Rate"]])
```

Now that I've trained the model and created the clusters, I want to look at all the counties in cluster0 and cluster1! I've also made a list of urban counties and rural counties in California just to compare results and see how they compare to my list that I made based on external research and California databases.

```python
urban = {
    "Alameda County", "Butte County", "Contra Costa County", "El Dorado County",
    "Fresno County", "Imperial County", "Kern County", "Kings County",
    "Los Angeles County", "Marin County", "Merced County", "Monterey County",
    "Napa County", "Orange County", "Placer County", "Riverside County",
    "Sacramento County", "San Benito County", "San Bernardino County",
    "San Diego County", "San Francisco County", "San Joaquin County",
    "San Luis Obispo County", "San Mateo County", "Santa Barbara County",
    "Santa Clara County", "Santa Cruz County", "Shasta County", "Solano County",
    "Sonoma County", "Stanislaus County", "Sutter County", "Tulare County",
    "Ventura County", "Yolo County", "Yuba County",
}

rural = {
    "Alpine County", "Amador County", "Calaveras County", "Colusa County",
    "Del Norte County", "Glenn County", "Humboldt County", "Inyo County",
    "Lake County", "Lassen County", "Madera County", "Mariposa County",
    "Mendocino County", "Modoc County", "Mono County", "Nevada County",
    "Plumas County", "Sierra County", "Siskiyou County", "Tehama County",
    "Trinity County", "Tuolumne County",
}
```

```python
cluster0 = county_means[county_means["Cluster"] == 0]
cluster0
```

```python
cluster1 = county_means[county_means["Cluster"] == 1]
cluster1
```

Now that I've seen what counties go into what cluster and have a list of the true urban and rural counties from external research and California databases, I can compare how accurate the clusters are. I can see what counties got misidentified and what cluster seems to misidentify more.

```python
def true_label(county):
    if county in urban:
        return 0   
    elif county in rural:
        return 1   
    else:
        return None  

county_means["TrueLabel"] = county_means["Area Name"].apply(true_label)

mismatches = county_means[county_means["Cluster"] != county_means["TrueLabel"]]

mismatches[["Area Name", "Cluster", "TrueLabel"]]
```

Let's look at some numeric values; we want to see how many counties were actually wrongly identified in Cluster 1 and Cluster 0.

```python
wrong_per_cluster = mismatches["Cluster"].value_counts()
wrong_per_cluster
```

Now that we have some numeric values, we can also look into the percentage of how accurate the clusters were. We will follow the formula of 1 minus the amount wrong in each cluster divided by the length of each cluster.

```python
1 - (13/len(cluster1))
```

```python
1 - (9 / len(cluster0))
```
